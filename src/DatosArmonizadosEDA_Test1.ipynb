```python
# EXOPLANET EDA - NASA KEPLER / K2 / TESS
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import missingno as msno
from pathlib import Path

# Configuración visual
sns.set(style="whitegrid", palette="mako", font_scale=1.1)
plt.rcParams["figure.figsize"] = (8,5)

# Ruta de tu dataset
DATA_FP = Path("/home/luna/Desktop/AWorldAwayLuna/FilteredData/Harmonized_Intersection.csv")

# Leer los datos
df = pd.read_csv(DATA_FP)
print(f"✅ Dataset cargado correctamente: {df.shape[0]:,} filas × {df.shape[1]} columnas")

# Mostrar primeras filas
df.head()
```

```python
# Información general
df.info()

# Porcentaje de nulos por columna
nulls = df.isna().mean().sort_values(ascending=False) * 100
nulls.to_frame("Porcentaje_Nulos (%)")
```

```python
# Mapa de calor de nulos
msno.heatmap(df, cmap="viridis")
plt.title("Mapa de calor de valores nulos", fontsize=14)
plt.show()

# Matriz de nulos por columna
msno.bar(df, color="teal")
plt.title("Cantidad de valores nulos por variable", fontsize=14)
plt.show()
```

```python
df.describe().T.style.background_gradient(cmap="mako")
```

```python
cols_to_check = [
    'orbital_period_days','transit_duration_hours','transit_depth_ppm',
    'planet_radius_earth','equilibrium_temperature_K','insolation_flux_Earth',
    'stellar_radius_solar','stellar_temperature_K'
]

for c in cols_to_check:
    if c in df.columns:
        neg = (df[c] < 0).sum()
        zeros = (df[c] == 0).sum()
        print(f"{c:25s} → Negativos: {neg:4d} | Ceros: {zeros:4d} | Rango: [{df[c].min():.3g}, {df[c].max():.3g}]")
```

```python
num_cols = [
    'orbital_period_days','transit_duration_hours','transit_depth_ppm',
    'planet_radius_earth','stellar_radius_solar','stellar_temperature_K'
]

for col in num_cols:
    plt.figure(figsize=(8,4))
    sns.histplot(df[col], kde=True, bins=50)
    plt.title(f"Distribución de {col}")
    plt.xlabel(col)
    plt.ylabel("Frecuencia")
    plt.show()
```

```python
for col in num_cols:
    plt.figure(figsize=(8,5))
    sns.boxplot(data=df, x="source_mission", y=col)
    plt.title(f"{col} — Comparación entre misiones")
    plt.show()
```

```python
plt.figure(figsize=(10,8))
corr = df[num_cols].corr()
sns.heatmap(corr, annot=True, cmap="mako", fmt=".2f", linewidths=0.5)
plt.title("Matriz de correlaciones (variables numéricas)", fontsize=14)
plt.show()
```

```python
fig, ax = plt.subplots(1,2, figsize=(12,5))

sns.countplot(data=df, x="final_disposition", order=["CONFIRMED","CANDIDATE","FALSE_POSITIVE"], ax=ax[0], palette="viridis")
ax[0].set_title("Distribución global de clases")
ax[0].set_xlabel("Clase")
ax[0].set_ylabel("Cantidad")

sns.countplot(data=df, x="source_mission", hue="final_disposition", ax=ax[1], palette="viridis")
ax[1].set_title("Distribución de clases por misión")
ax[1].set_xlabel("Misión")
ax[1].legend(title="Clase", bbox_to_anchor=(1.05,1), loc="upper left")

plt.tight_layout()
plt.show()
```

```python
sns.pairplot(
    df,
    vars=['planet_radius_earth','transit_depth_ppm','orbital_period_days','stellar_radius_solar'],
    hue='final_disposition',
    palette='viridis',
    diag_kind='kde',
    plot_kws={'alpha':0.5}
)
plt.suptitle("Relaciones entre variables físicas y clase", y=1.02)
plt.show()
```

```python
summary = df.groupby("source_mission")["final_disposition"].value_counts(normalize=True).unstack().fillna(0)*100
summary.style.background_gradient(cmap="mako").format("{:.2f}%")
```
